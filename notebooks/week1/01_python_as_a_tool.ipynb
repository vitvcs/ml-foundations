Before working with AI models, it is essential to fully trust the basic

# Question

Can I correctly implement basic statistical operations using only pure Python?

# Hypothesis

Implementing these operations manually will improve my understanding of how
scientific libraries work internally.

```python
data = [10,12,9,15,11]

min_num = None
max_num = None
avg_num = None
sum_num = 0
count_num = 0

for value in data:
    if min_num == None:
        min_num = value
    elif min_num > value:
        min_num = value
    if max_num == None:
        max_num = value
    elif max_num < value:
        max_num = value
    sum_num += value
    count_num += 1
avg_num = sum_num/count_num

print('Lowest value:',min_num)
print('Largest value:',max_num)
print('Average:',avg_num)
```
